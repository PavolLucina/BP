```python
import papermill as pm
import nbformat
import pandas as pd
import os

SEEDS = [0, 1, 2, 3, 4]


MEGADESCRIPTOR_VERSIONS = ['T-224'  , 'S-224', 'B-224', 'L-384']
BASE_OUTPUT_DIR = "outputs_wb_seg"
os.makedirs(BASE_OUTPUT_DIR, exist_ok=True)

results = []

MODES = ["", "wb_"]  # segmentation + grayscale segmentation
```

```python
for mode in MODES:
    for megadescriptor_version in MEGADESCRIPTOR_VERSIONS:
        for seed in SEEDS:

            run_name = f"{mode}seg_{megadescriptor_version}_seed{seed}"
            output_path = f"{BASE_OUTPUT_DIR}/test_run_{run_name}.ipynb"

            print(f"Running: mode={mode} | megadescriptor_version={megadescriptor_version} | seed={seed}")

            pm.execute_notebook(
                f"{BASE_OUTPUT_DIR}/test_seg.ipynb",
                output_path,
                parameters=dict(
                    megadescriptor_version=megadescriptor_version,
                    detection="_seg",   
                    mode=mode,
                    seed=seed,
                    query_ratio=0.2
                )
            )

            # read result
            nb = nbformat.read(output_path, as_version=4)
            output_cell = nb.cells[-1]

            output_text = output_cell["outputs"][0]["data"]["text/plain"]
            d = eval(output_text)

            results.append({
                "mode": mode if mode != "" else "rgb",
                "megadescriptor_version": d["megadescriptor_version"],
                "seed": d["seed"],
                "accuracy": float(d["accuracy"])
            })
```

```python
import pandas as pd

df = pd.DataFrame(results)
df["mode"] = df["mode"].replace({"": "rgb_segmentation", "wb_": "bw_segmentation"})
summary = (
    df.groupby(["mode", "megadescriptor_version"])
    .agg(
        mean_accuracy=("accuracy", "mean"),
        std_accuracy=("accuracy", "std")
    )
    .reset_index()
    .sort_values(["mode", "mean_accuracy"], ascending=[True, False])
)

display(summary)
```

```python
df.to_csv("segmentation_test_results.csv", index=False)
```
