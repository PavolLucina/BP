```python
import papermill as pm
import json
import ast
import nbformat

with open("outputs/test_run.ipynb") as f:
    nb = json.load(f)

for cell in nb["cells"]:
    for output in cell.get("outputs", []):
        if "text" in output:
            try:
                data = ast.literal_eval("".join(output["text"]))
                if "accuracy" in data:
                    print(data["accuracy"])
            except:
                pass
```

```python
import numpy as np
import torch
import pandas as pd
from sklearn.model_selection import StratifiedShuffleSplit

SEEDS = [0, 1, 2, 3, 4]

DATASET_VERSIONS = [
    "",
    "_detected",
    "_detected_manual"
]

MODEL_VERSIONS = [
    'T-224','S-224', 'B-224', 'L-384'
]



results = []
i = 0
for dataset_version in DATASET_VERSIONS:
    for model_version in MODEL_VERSIONS:
        for seed in SEEDS:
            pm.execute_notebook(
                "test_saved_models.ipynb",
                f"outputs/test_run_{dataset_version}_{model_version}_{seed}.ipynb",
                parameters=dict(
                        megadescriptor_version=model_version,
                        detection=dataset_version,
                        query_ratio=0.2,
                        seed=seed
                )
            )

            output_cell = nbformat.read(f"outputs/test_run_{dataset_version}_{model_version}_{seed}.ipynb", as_version=4).cells[-1]
            output_text = output_cell["outputs"][0]["data"]["text/plain"]
            d = eval(output_text)
            accuracy = float(d['accuracy'])

            results.append({
                "dataset": dataset_version,
                "model": model_version,
                "seed": seed,
                "accuracy": accuracy
            })
            i += 1


df = pd.DataFrame(results)

# aggregate
summary = (
    df
    .groupby(["dataset", "model"])
    .agg(
        mean_accuracy=("accuracy", "mean"),
        std_accuracy=("accuracy", "std")
    )
    .reset_index()
)

display(summary.sort_values("mean_accuracy", ascending=False))
```
